```python
#VARIABLES

name = "Alice"
age = 25
height = 5.6

print("Name:", name)
print("Age:", age)
print("Height:", height)

name = 123  # Now 'name' holds an integer
print("New value of name:", name)

```

```python
#DATATYPES

# Integer
a = 10
print(type(a))


# Float
b = 3.14
print(type(b))


# String
c = "Python"
print(type(c))

# Boolean
d = True
print(type(d))

#LIST
e = [1, 2, 3, "four"]
print(type(e), e)

# Dictionary
f = {"name": "Alice", "age": 25}
print(type(f), f)
```
